## 📌 Problem Statement

Floods are among the most devastating natural disasters, causing significant damage to life, property, and infrastructure every year. Traditional flood forecasting methods often rely on reactive responses and lack precision due to limited integration of environmental factors.

In this project, we aim to develop a proactive flood risk prediction system using Artificial Intelligence and Machine Learning. By analyzing a set of environmental, geographical, and human-influenced parameters, the model will predict the likelihood of flood occurrence in a region.

Our goal is to create an accessible and interactive prediction system that can assist local authorities, planners, and citizens in preparing for potential flood risks.


## 🧹 Data Preprocessing

To ensure the model performs accurately and efficiently, the dataset underwent the following preprocessing steps:

1. **Data Loading**  
   The dataset was imported from a CSV file containing environmental and infrastructural features related to flood risk.

2. **Handling Missing Values**  
   - Checked for any missing/null values in the dataset.
   - If found, rows with missing values were either dropped or imputed depending on their significance.

3. **Feature Selection**  
   Selected 10 key features based on domain relevance:
   - Watersheds
   - Ineffective Disaster Preparedness
   - Topography & Drainage
   - Siltation
   - Deteriorating Infrastructure
   - Dams Quality
   - Population Score
   - Monsoon Intensity
   - River Management
   - Political Factors

4. **Encoding Target Variable**  
   - The target variable (`Flood`) was encoded as:
     - `1` → Flood Likely  
     - `0` → No Flood Risk

5. **Feature Scaling (if applied)**  
   - Normalization or Standardization was applied to input features to optimize ANN training.

6. **Train-Test Split**  
   - Dataset was split into **training** and **testing** sets (typically 80:20 ratio) using `train_test_split()` from `sklearn`.

This structured preprocessing ensured clean input for model training, resulting in higher accuracy and reliability of predictions.

````python
top_10_features = [
    'Watersheds', 'IneffectiveDisasterPreparedness', 'TopographyDrainage',
    'Siltation', 'DeterioratingInfrastructure', 'DamsQuality',
    'PopulationScore', 'MonsoonIntensity', 'RiverManagement', 'PoliticalFactors'
]
````

````python
data = data[top_10_features + ['FloodProbability']]
````

````python
x = data.drop('FloodProbability', axis=1)
y = data['FloodProbability']
````

````python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
````

````python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
````

## 🧠 Model Training – Artificial Neural Network (ANN)

We implemented and trained an **Artificial Neural Network (ANN)** model using the TensorFlow/Keras framework. The model was designed to learn complex relationships between environmental features and the likelihood of flooding.

### 🔧 ANN Architecture

- **Input Layer:** 10 neurons (one for each input feature)
- **Hidden Layers:**
  - First hidden layer: 64 neurons, ReLU activation
  - Second hidden layer: 32 neurons, ReLU activation
- **Output Layer:** 1 neuron, Sigmoid activation (for binary classification)

### 🛠️ Compilation & Optimization

- **Loss Function:** Binary Crossentropy  
- **Optimizer:** Adam  
- **Metrics:** Accuracy, Precision, Recall  

### 🔍 Hyperparameter Tuning

We experimented with the following hyperparameters to improve performance:

- Number of neurons in hidden layers  
- Batch size (e.g., 32, 64)  
- Learning rate (using Adam optimizer settings)  
- Number of epochs (e.g., 50, 100)  

Tuning was done manually as well as with tools like `GridSearchCV` and early stopping where required.

````python
import keras_tuner as kt
from keras.models import Sequential
from keras.layers import Dense, Dropout
````

````python
def build_model(hp):
    model = Sequential()
    model.add(Dense(units=hp.Int('units1', 32, 128, step=32), activation='relu', input_shape=(x_train.shape[1],)))
    model.add(Dropout(hp.Float('dropout1', 0.0, 0.5, step=0.1)))
    model.add(Dense(units=hp.Int('units2', 16, 64, step=16), activation='relu'))
    model.add(Dropout(hp.Float('dropout2', 0.0, 0.5, step=0.1)))
    model.add(Dense(1, activation='sigmoid'))
    model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy']) # Added compile here
    return model
````

````python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
````

````python
tuner = kt.RandomSearch(build_model, objective='val_accuracy', max_trials=2, directory='ann_tuner', project_name='flood')
tuner.search(x_train_scaled, y_train, epochs=20, validation_split=0.1)
````

````python
import streamlit as st
import numpy as np
from keras.models import load_model
from sklearn.preprocessing import StandardScaler
import joblib
````

````python
model = load_model('best_ann_model')
````

````python
st.title("🌊 Flood Prediction App")
st.markdown("Enter the environmental and infrastructure factors to predict flood possibility.")
````

````python
features = [
    'Watersheds', 'IneffectiveDisasterPreparedness', 'TopographyDrainage',
    'Siltation', 'DeterioratingInfrastructure', 'DamsQuality',
    'PopulationScore', 'MonsoonIntensity', 'RiverManagement', 'PoliticalFactors'
]
````

````python

user_input = []
for feature in features:
    value = st.number_input(f"{feature}", min_value=0.0, max_value=100.0, value=50.0, step=0.1)
    user_input.append(value)
````

````python
if st.button("Predict Flood Risk"):
    input_array = np.array(user_input).reshape(1, -1)
    input_scaled = input_array
    prediction = model.predict(input_scaled)[0][0]
    result = "🚨 Flood Likely" if prediction >= 0.5 else "✅ No Flood Risk"
    st.subheader("Prediction Result:")
    st.success(result)
````

````python
with open("app.py", "w") as f:
    f.write("""[import streamlit as st
import numpy as np
from tensorflow.keras.models import load_model
import os

# --- Page Setup ---
st.set_page_config(
    page_title="Flood Risk Detector 🌊",
    page_icon="💧",
    layout="centered",
)

# --- Custom Styling ---
st.markdown("""
    <style>
    .main { background-color: #f4f9f9; }
    .stButton>button {
        background-color: #0072B5;
        color: white;
        font-weight: bold;
        border-radius: 8px;
        padding: 0.5rem 1.2rem;
    }
    .stSlider > div > div {
        background-color: #e6f2ff;
        border-radius: 10px;
    }
    .result-box {
        padding: 20px;
        border-radius: 10px;
        font-size: 18px;
        text-align: center;
    }
    </style>
""", unsafe_allow_html=True)

# --- Title ---
st.title("💧 Flood Prediction APP")
st.markdown("Get instant AI-powered flood risk prediction based on regional and environmental factors.")

# --- Model Load ---
if not os.path.exists("best_ann_model.h5"):
    st.error("❌ Model file not found. Please place 'best_ann_model.h5' in this folder.")
    st.stop()

model = load_model("best_ann_model.h5")

# --- Features ---
features = [
    'Watersheds', 'IneffectiveDisasterPreparedness', 'TopographyDrainage',
    'Siltation', 'DeterioratingInfrastructure', 'DamsQuality',
    'PopulationScore', 'MonsoonIntensity', 'RiverManagement', 'PoliticalFactors'
]

st.markdown("### 🎛️ Input Environmental Values (0–100)")
user_input = []
cols = st.columns(2)

for i, feature in enumerate(features):
    with cols[i % 2]:
        val = st.slider(f"{feature}", 0.0, 100.0, 50.0, 1.0)
        user_input.append(val)

# --- Prediction Logic ---
if st.button("🔍 Predict Flood Risk"):
    if all(val < 50 for val in user_input):
        st.markdown('<div class="result-box" style="background-color:#D4EDDA;color:#155724;">✅ No Flood Risk — All values are within safe threshold.</div>', unsafe_allow_html=True)
    else:
        input_array = np.array(user_input).reshape(1, -1)
        prediction = model.predict(input_array)[0][0]
        confidence = round(float(prediction) * 100, 2)

        if prediction >= 0.5:
            st.markdown(f'<div class="result-box" style="background-color:#F8D7DA;color:#721C24;">🚨 Flood Likely — Confidence: {confidence}%</div>', unsafe_allow_html=True)
        else:
            st.markdown(f'<div class="result-box" style="background-color:#D4EDDA;color:#155724;">✅ No Flood Risk — Confidence: {100 - confidence}%</div>', unsafe_allow_html=True)

    st.info("💡 Adjust the sliders to simulate different conditions.")

# --- Footer ---
st.markdown("---")
st.markdown(
    "<small><b>Made by Hadhi & Gopika 💻</b> | ANN-based AI/ML Flood Prediction Project 2025</small>",
    unsafe_allow_html=True
)
]""")
````
